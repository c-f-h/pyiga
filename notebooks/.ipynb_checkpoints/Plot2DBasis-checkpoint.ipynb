```python
%matplotlib inline
%load_ext autoreload
%autoreload 2

import pyiga

import scipy
import numpy as np
import matplotlib.pyplot as plt

from scipy.sparse import csc_matrix, csr_matrix, coo_matrix
from scipy.sparse.linalg import aslinearoperator

np.set_printoptions(linewidth=100000)
np.set_printoptions(precision=5)

from pyiga import bspline, assemble, vform, geometry, vis, solvers, topology, bspline_cy
```

```python
geo = geometry.quarter_annulus(1.0,2)
```

```python
kvs = 2* (bspline.make_knots(3,0,1,5),)
```

```python
kvs[0].numdofs
```

```python
phi = bspline.BSplineFunc(kvs, np.eye(1,64,27).ravel())
```

```python
# parameter domain
N = 4
nu, nv = 6+(2**N-1)*5, 6+(2**N-1)*5
u = np.linspace(0, 1, nu)
v = np.linspace(0, 1, nv)

U, V = np.meshgrid(u, v, indexing="ij")
Z = phi(u,v)
```
