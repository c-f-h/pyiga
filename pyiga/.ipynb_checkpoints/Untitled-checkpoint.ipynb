```python
def make_bubble(nodes):
    """
    Build a global piecewise tensor-product bubble function
    on a tensor grid induced by `nodes`.

    Parameters
    ----------
    nodes : array_like
        Increasing nodes in [0,1]

    Returns
    -------
    b : callable
        b(x,y): evaluates the bubble function
    """
    nodes = np.asarray(nodes, dtype=float)

    if np.any(np.diff(nodes) <= 0):
        raise ValueError("nodes must be strictly increasing")
    h = np.diff(nodes)

    def phi(t):
        t = np.asarray(t)
        
        # interval index containing t
        idx = np.searchsorted(nodes, t, side='right') - 1
        
        # outside [0,1]
        valid = (idx >= 0) & (idx < len(h))
        out = np.zeros_like(t, dtype=float)
        k = idx[valid]
        xi = (t[valid] - nodes[k]) / h[k]
        out[valid] = 4*xi * (1.0 - xi)
        return out

    def b(x, y):
        return phi(x) * phi(y)

    return b
```

```python
def P(n, x):
    if n == 0:
        return 1.0
    if n == 1:
        return x

    return ((2*n - 1)*x*P(n-1, x) - (n-1)*P(n-2, x)) / n

def psi(p,sd):
    if sd == 0:
        return lambda t:(1-t)*(-1)**(p+1)*P(p+1,2*t-1)
    elif sd==1:
        return lambda t:t*P(p+1,2*t-1)

def make_bubble(nodes, S, p, ax, sd):
    nodes = np.asarray(nodes, dtype=float)

    if np.any(np.diff(nodes) <= 0):
        raise ValueError("nodes must be strictly increasing")

    h = np.diff(nodes)
    psi_p = psi(p,sd)     # create once

    def phi(t):
        t = np.asarray(t)

        idx = np.searchsorted(nodes, t, side='right') - 1

        valid = (idx >= 0) & (idx < len(h))
        out = np.zeros_like(t, dtype=float)

        k = idx[valid]
        xi = (t[valid] - nodes[k]) / h[k]

        out[valid] = 4 * xi * (1 - xi) * S(t[valid])

        return out

    if ax==0:
        if sd==0:
            def b(x, y):
                y = np.asarray(y)
                return np.where(y <= h[0],psi_p(y/h[0]) * phi(x),0.0)
        elif sd==1:
            def b(x, y):
                y = np.asarray(y)
                return np.where(1-y <= h[0],psi_p((y-nodes[-2])/h[-1]) * phi(x),0.0)

    elif ax==1:
        if sd==0:
            def b(x, y):
                y = np.asarray(y)
                return np.where(x <= h[0],psi_p(x/h[0]) * phi(y),0.0)
        elif sd==1:
            def b(x, y):
                y = np.asarray(y)
                return np.where(1-x <= h[0],psi_p((x-nodes[-2])/h[-1]) * phi(y),0.0)

    return b
```
